# Matplot Example Workflow


```python
## 0. import matplotlib and get it ready for plotting in Jupyter
%matplotlib inline
import matplotlib.pyplot as plt

## 1. Prepare data
x = [1, 2, 3, 4,]
y = [11, 22, 33, 44]

## 2. Setup plot
fix, ax = plt.subplots(figsize=(20, 20))  # width and height of plot

## 3. PLot data
ax.plot(x, y)

## 4. Customize plot
ax.set(title="Simple Plot", xlabel="x-axis", ylabel="y-axis")

## 5. Save and show (you save the figure)
fig.savefig('images/sample-plot.png')
```

```python
# Make a plot from a dictionary
nut_butter_prices = {"Almond butter": 10, 
                     "Peanut butter": 8,
                     "Cashew butter": 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax.set(title="ButterStore", ylabel="Price ($)", xlabel="Items");

```

```python
## Make some data for histogram and plot it
%matplotlib inline 
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 

x = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(x);
```

```python
## Use subplots to create multiple sub-plots within

nut_butter_prices = {"Almond butter": 10, 
                     "Peanut butter": 8,
                     "Cashew butter": 12}


### Option 1 - tuple of tuples
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots( nrows=2, 
                                             ncols=2, 
                                             figsize= (10,5))

# Plot to each different axis 
ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10));
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax4.hist(x); 
```

```python
### Option 2 for Subplot - List index way
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,5))

# Plot to each different index
ax[0, 0].plot(x, x/2);
ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1, 1].hist(x); 

```

```python
# Another dataset
heart_disease = pd.read_csv("heart-disease.csv")
heart_disease.head()
```

```python
# Create a histogram of age
heart_disease["age"].plot.hist(bins=10)
```

```python
heart_disease.head()
```

```python
heart_disease.plot.hist(figsize=(10,30), subplots=True);
```

```python
## When plotting more advance use the matplotlib OO method
over_50 = heart_disease[heart_disease["age"] > 50]
over_50
```

```python
#PyPlot method
over_50.plot(kind='scatter', 
             x="age",
             y="chol",
             c='target');
```

```python
# OO Method (ax is Axes)
fig, ax  = plt.subplots(figsize=(10,6))
over_50.plot(kind='scatter',
             x="age",
             y="chol",
             c="target",
             ax=ax);

ax.set_xlim([45, 100])
```
